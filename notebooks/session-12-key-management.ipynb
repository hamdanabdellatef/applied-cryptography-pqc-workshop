```python
"""Interactive teaching experiments. Cryptographic operations and policy models are labelled.

Uses the lesson's already-executed namespace; does not contact external services.
"""
import hashlib
from html import escape

TEACHING_CASES = {
    7: ['valid', 'wrong-hostname', 'unknown-root', 'mutual-tls', 'missing-client', 'wrong-client-purpose'],
    8: ['archive', 'research', 'telemetry'],
    9: ['valid', 'altered-kem-ciphertext', 'truncated-kem-ciphertext', 'wrong-recipient', 'changed-aad'],
    10: ['valid', 'zero-pq-contribution', 'changed-transcript', 'missing-contribution', 'downgrade', 'unauthenticated'],
    11: ['valid', 'changed-message', 'changed-context', 'wrong-key', 'truncated-signature', 'old-version'],
    12: ['valid', 'wrong-kek', 'changed-context', 'rewrap', 'stolen-old-backup'],
    13: ['valid', 'wrong-tenant', 'operator-decrypt', 'compromised-reader', 'revoked-grant'],
    14: ['storage-reader', 'worker-acme', 'recovery-officer', 'broaden-worker', 'unknown-principal'],
    15: ['valid', 'legacy-read', 'legacy-write', 'unknown-profile', 'backup-not-ready', 'retire'],
    16: ['baseline', 'verifier-delay', 'transport-delay', 'cycle'],
}


def teaching_step(name, explanation, artifacts=None, status='OBSERVE'):
    return {'step':name, 'status':status, 'explanation':explanation, 'artifacts':artifacts or {}}


def teaching_steps(session, ns, case='valid', parameter=10):
    """Lazy generator: Next step executes the next operation, not just reveals text."""
    if session not in TEACHING_CASES or case not in TEACHING_CASES[session]:
        raise ValueError('Unknown session or case')
    def step(name, explanation, artifacts=None, status='OBSERVE'):
        return teaching_step(name, explanation, artifacts, status)
    if session == 7:
        import ssl
        options = {'wrong-hostname':{'hostname':'other.test'}, 'unknown-root':{'trust_root':False},
                   'mutual-tls':{'mtls':True}, 'missing-client':{'mtls':True,'send_client':False},
                   'wrong-client-purpose':{'mtls':True,'client_wrong_eku':True}}.get(case,{})
        yield step('Configure endpoints', 'Real TLS 1.3 via MemoryBIO. Verification remains enabled.', {'options':(options,False)})
        try:
            result = ns['tls_trial'](**options)
        except ssl.SSLError as error:
            yield step('TLS handshake', str(error), status='REJECT'); return
        yield step('TLS handshake and record', 'This is a whole real handshake, not a packet-by-packet trace.', {'negotiated/result':(result,False)},'PASS')
        yield step('Interpret result', 'Cipher suite does not identify the key-exchange group. mTLS identity still needs application authorization.', status='LIMIT')
    elif session == 8:
        asset = {'archive':(20,5),'research':(10,3),'telemetry':(1,2)}[case]
        lifetime, migration = asset
        yield step('Assumptions', 'Hypothetical years, not a forecast.', {'L, M, H':((lifetime,migration,parameter),False)})
        gap = ns['planning_gap'](lifetime,migration,parameter)
        yield step('Compute L + M - H', f'{lifetime} + {migration} - {parameter} = {gap}', status='GAP' if gap>0 else 'NO POSITIVE GAP')
        yield step('Bound the claim', 'Zero/negative gap does not prove safety, remove existing recordings or address endpoint compromise.', status='LIMIT')
    elif session == 9:
        from cryptography.hazmat.primitives.asymmetric.mlkem import MLKEM768PrivateKey
        from cryptography.hazmat.primitives.ciphers.aead import AESGCM
        from cryptography.exceptions import InvalidTag
        import secrets
        bob = ns['bob_kem']
        secret, ct = bob.public_key().encapsulate()
        yield step('Encapsulate to Bob', 'Inspect public encapsulation key, KEM ciphertext and locally generated secret.',
                   {'public key':(bob.public_key().public_bytes_raw(),False),'KEM ciphertext':(ct,False),'sender secret':(secret,True)})
        received = ct[:-1] if case=='truncated-kem-ciphertext' else bytes([ct[0]^1])+ct[1:] if case=='altered-kem-ciphertext' else ct
        recipient = MLKEM768PrivateKey.generate() if case=='wrong-recipient' else bob
        try:
            recovered = recipient.decapsulate(received)
        except ValueError:
            yield step('Decapsulate', 'Wrong-length ciphertext rejected as malformed.',status='REJECT'); return
        yield step('Decapsulate', f'Returned {len(recovered)} bytes; equal secrets = {secret == recovered}. Returning bytes is not peer authentication.', {'receiver secret':(recovered,True)})
        transcript = b'panel:ML-KEM-768:v1|' + bob.public_key().public_bytes_raw() + ct
        sender_key, receiver_key = (ns['derive_day2'](s,transcript) for s in (secret,recovered))
        yield step('Derive record keys', 'Hold original transcript fixed to isolate secret mismatch; this is a teaching schedule.', {'sender key':(sender_key,True),'receiver key':(receiver_key,True)})
        nonce, aad = secrets.token_bytes(12), b'invoice-panel:v1'
        record = AESGCM(sender_key).encrypt(nonce,b'synthetic invoice',aad)
        try:
            plaintext = AESGCM(receiver_key).decrypt(nonce,record,aad+b'!' if case=='changed-aad' else aad)
        except InvalidTag:
            yield step('Authenticate record', 'AEAD rejected; no plaintext released.',{'AEAD record':(record,False)},'REJECT'); return
        yield step('Authenticate record', 'Round trip succeeds. Identity and replay protection remain protocol requirements.',{'plaintext':(plaintext,False)},'PASS')
    elif session == 10:
        selected = 'classical-only' if case=='downgrade' else 'approved-hybrid-profile'
        yield step('Negotiation inputs', 'Application policy model, not authenticated wire negotiation.',{'selected profile':(selected,False)})
        try:
            ns['require_hybrid'](selected,case!='unauthenticated')
        except ValueError as error:
            yield step('Minimum policy', str(error),status='REJECT'); return
        yield step('Contributions', 'Fixed-size classical and PQ secrets from the lesson.', {'classical':(ns['zx_a'],True),'PQ':(ns['zp_a'],True)})
        pq = b'' if case=='missing-contribution' else bytes(32) if case=='zero-pq-contribution' else ns['zp_a']
        context = ns['transcript'] + (b'changed' if case=='changed-transcript' else b'')
        try:
            key = ns['teaching_combiner'](ns['zx_a'],pq,context)
        except ValueError as error:
            yield step('Combiner', str(error),status='REJECT'); return
        yield step('Compare output', f'Matches baseline = {key == ns["key_a"]}. Sensitivity is not a robust-combiner proof; zeros are not a fallback.', {'derived key':(key,True)},'OBSERVE')
    elif session == 11:
        from cryptography.hazmat.primitives.asymmetric.mldsa import MLDSA65PrivateKey
        from cryptography.exceptions import InvalidSignature
        message = ns['message'].replace(b'version=9',b'version=8') if case=='old-version' else ns['message']
        signature = ns['signer'].sign(message,ns['context'])
        yield step('Sign exact release bytes', 'This signature binds message and context; public key trust is configured separately.', {'message':(message,False),'context':(ns['context'],False),'public key':(ns['public'].public_bytes_raw(),False),'signature':(signature,False)})
        public = MLDSA65PrivateKey.generate().public_key() if case=='wrong-key' else ns['public']
        try:
            public.verify(signature[:-1] if case=='truncated-signature' else signature,
                          message+b'!' if case=='changed-message' else message,
                          b'other-purpose' if case=='changed-context' else ns['context'])
        except InvalidSignature:
            yield step('Verify ML-DSA', 'Cryptographic verification rejected the altered input.',status='REJECT'); return
        yield step('Verify ML-DSA', 'Signature is valid relative to this configured public key.',status='PASS')
        yield step('Release policy', 'Teaching policy requires exact version 9 for this fixed-format message; a valid old signature is insufficient.', status='REJECT' if case=='old-version' else 'PASS')
    elif session == 12:
        from cryptography.hazmat.primitives.keywrap import aes_key_unwrap, aes_key_wrap, InvalidUnwrap
        from cryptography.hazmat.primitives.ciphers.aead import AESGCM
        from cryptography.exceptions import InvalidTag
        yield step('Inspect envelope', 'DEK and KEKs are disposable local secrets. Ciphertext and wrapped DEK are separate objects.', {'wrapped DEK':(ns['wrapped_old'],False),'ciphertext':(ns['ciphertext'],False),'DEK':(ns['dek'],True),'old KEK':(ns['kek_old'],True),'new KEK':(ns['kek_new'],True)})
        try:
            dek = aes_key_unwrap(ns['kek_new'] if case=='wrong-kek' else ns['kek_old'],ns['wrapped_old'])
        except InvalidUnwrap:
            yield step('Unwrap DEK', 'Wrong KEK rejected by AES Key Wrap.',status='REJECT'); return
        yield step('Unwrap DEK', 'Recovered DEK matches original.',{'recovered DEK':(dek,True)},'PASS')
        if case in ('rewrap','stolen-old-backup'):
            wrapped = aes_key_wrap(ns['kek_new'],dek)
            yield step('Rotate wrapper', 'New KEK wraps the same DEK. Bulk ciphertext is unchanged.',{'new wrapped DEK':(wrapped,False)})
        try:
            plain = AESGCM(dek).decrypt(ns['nonce'],ns['ciphertext'],ns['context']+(b'!' if case=='changed-context' else b''))
        except InvalidTag:
            yield step('Authenticate data', 'Changed expected context rejected.',status='REJECT'); return
        yield step('Decrypt historical copy' if case=='stolen-old-backup' else 'Decrypt data',
                   'Old KEK plus old wrapped copy still reveals plaintext after rewrap.' if case=='stolen-old-backup' else 'AEAD accepted the record and context.', {'plaintext':(plain,False)},'EXPOSED' if case=='stolen-old-backup' else 'PASS')
    elif session == 13:
        principal = 'key-operator' if case=='operator-decrypt' else 'archive-reader'
        tenant = 'tenant-other' if case=='wrong-tenant' else 'tenant-acme'
        yield step('Authenticated caller assumption', 'Policy simulation only. Compromise does not change the authenticated identity string.',{'principal':(principal,False),'tenant':(tenant,False)})
        grant = ('archive-reader','decrypt','tenant-acme')
        saved = set(ns['grants'])
        try:
            if case=='revoked-grant': ns['grants'].discard(grant)
            allowed = ns['authorize'](principal,'decrypt',tenant,'interactive-teaching-request')
        finally:
            ns['grants'].clear(); ns['grants'].update(saved)
        yield step('Authorization and audit', 'Temporary policy edits restored after this trial.',{'latest audit event':(ns['audit'][-1],False)},'PASS' if allowed else 'REJECT')
        if allowed:
            yield step('Interpret allowed use', 'A compromised authorized reader can still request its scoped plaintext; key non-exportability does not prevent this.',status='EXPOSED' if case=='compromised-reader' else 'LIMIT')
    elif session == 14:
        model = {p:set(a) for p,a in ns['plaintext_access'].items()}
        principal = 'worker-acme' if case=='broaden-worker' else case
        if case=='broaden-worker': model[principal].add('document-other')
        yield step('Declared design', 'Explicit assumption model, not discovered permissions or an escalation analysis.',{'principal':(principal,False),'access map':({p:sorted(a) for p,a in model.items()},False)})
        try:
            exposed = ns['exposed_after_compromise']([principal],model)
        except ValueError as error:
            yield step('Evaluate scope',str(error),status='UNKNOWN'); return
        yield step('Modeled plaintext reach', 'An empty set does not rule out metadata theft, deletion or escalation.',{'exposed assets':(sorted(exposed),False)},'EXPOSED' if exposed else 'NO MODELED PLAINTEXT')
    elif session == 15:
        profile = 'record-v1' if case in ('legacy-read','legacy-write') else 'unknown' if case=='unknown-profile' else 'record-v2'
        operation = 'read' if case in ('legacy-read','unknown-profile') else 'write'
        yield step('Trusted policy', 'Profile routing model; not authenticated format negotiation.',{'policy':(ns['policy'],False),'request':((operation,profile),False)})
        try:
            provider = ns['select_profile'](profile,operation,ns['policy'])
        except ValueError as error:
            yield step('Select provider',str(error),status='REJECT'); return
        yield step('Select provider','Allowed under explicit read/write policy.',{'provider':(provider,False)},'PASS')
        evidence = dict(ns['evidence'])
        if case=='backup-not-ready': evidence['backups_recoverable']=False
        allowed = ns['can_retire'](evidence)
        yield step('Retirement gate', 'Boolean assertions require actual evidence and owner review.',{'evidence':(evidence,False)},'READY' if allowed else 'BLOCKED')
        if case=='retire':
            try: ns['select_profile']('record-v1','read',ns['retired_policy'])
            except ValueError:
                yield step('Retired reader', 'Old reads now reject; restore dependencies must have been resolved.',status='REJECT')
    elif session == 16:
        from graphlib import TopologicalSorter, CycleError
        dependencies = {k:set(v) for k,v in ns['dependencies'].items()}
        durations = dict(ns['duration'])
        if case=='verifier-delay': durations['verifier upgrade'] += parameter
        if case=='transport-delay': durations['transport pilot'] += parameter
        if case=='cycle': dependencies['inventory'].add('rollout')
        yield step('Planning assumptions', 'Invented months, unlimited parallel capacity; no quantum arrival forecast.', {'durations':(durations,False),'dependencies':({k:sorted(v) for k,v in dependencies.items()},False)})
        try: order = list(TopologicalSorter(dependencies).static_order())
        except CycleError:
            yield step('Order dependencies','Circular prerequisites prevent a valid order.',status='REJECT'); return
        finish = {}
        for task in order:
            start = max((finish[p] for p in dependencies[task]),default=0)
            finish[task] = start+durations[task]
            yield step(task,f'Start {start}; duration {durations[task]}; finish {finish[task]} months.',status='CALCULATED')
        yield step('Rollout finish',f'{finish["rollout"]} months under these assumptions. Identify the controlling prerequisite before changing budget or staffing.',{'finish times':(finish,False)},'PLAN')


def print_teaching_trace(session, namespace, case, parameter=10):
    rows = list(teaching_steps(session,namespace,case,parameter))
    for row in rows:
        print(row['step'], '|', row['status'], '|', row['explanation'])
    return rows


def teaching_panel(session, namespace):
    import ipywidgets as widgets
    from IPython.display import display, HTML
    case = widgets.Dropdown(options=TEACHING_CASES[session],description='Scenario:')
    parameter = widgets.IntSlider(value=10,min=0,max=40,description='H (years):' if session==8 else 'Delay months:')
    if session not in (8,16): parameter.layout.display='none'
    secret = widgets.Checkbox(value=False,description='Reveal disposable secrets',indent=False)
    advance = widgets.Button(description='Next step',button_style='info')
    reset_button = widgets.Button(description='Reset experiment')
    all_button = widgets.Button(description='Run all steps')
    inspect_button = widgets.Button(description='Print revealed objects')
    output, inspection = widgets.Output(), widgets.Output()
    state={'rows':[],'iterator':None}

    def render():
        output.clear_output(wait=True)
        with output:
            style='border:1px solid #839aaa;padding:8px;white-space:normal;overflow-wrap:anywhere;vertical-align:top'
            rows=''.join('<tr>'+''.join(f'<td style="{style}">{escape(str(row[k]))}</td>' for k in ('step','status','explanation'))+'</tr>' for row in state['rows'])
            display(HTML(f'<table style="width:100%;table-layout:fixed;border-collapse:collapse"><tr><th>Step</th><th>Observation</th><th>Explanation</th></tr>{rows}</table>'))
            print('Predict before Next step. Completed traces do not automatically imply security acceptance.')

    def reset(_=None):
        state['rows']=[]
        state['iterator']=iter(teaching_steps(session,namespace,case.value,parameter.value))
        advance.disabled=False
        secret.value=False
        inspection.clear_output()
        render()

    def next_step(_=None):
        inspection.clear_output()
        try: state['rows'].append(next(state['iterator']))
        except StopIteration: advance.disabled=True
        render()

    def run_all(_):
        state['rows'].extend(state['iterator'])
        advance.disabled=True
        inspection.clear_output()
        render()

    def inspect(_):
        inspection.clear_output(wait=True)
        with inspection:
            for row in state['rows']:
                for label,(value,sensitive) in row['artifacts'].items():
                    print('\n'+row['step']+' / '+label)
                    if sensitive and not secret.value:
                        print('[hidden disposable secret; enable reveal and print again]'); continue
                    if isinstance(value,bytes):
                        print('Length:',len(value),'bytes; SHA-256:',hashlib.sha256(value).hexdigest())
                        print('Hex:',value.hex())
                        if not sensitive and label in ('message','context','plaintext'): print('Bytes:',repr(value))
                    else: print(value)
    advance.on_click(next_step); reset_button.on_click(reset)
    all_button.on_click(run_all); inspect_button.on_click(inspect)
    case.observe(reset,names='value'); parameter.observe(reset,names='value')
    secret.observe(lambda _: inspection.clear_output(),names='value')
    display(widgets.VBox([widgets.HBox([case,parameter]),widgets.HBox([advance,all_button,reset_button]),output,secret,inspect_button,inspection]))
    reset()
    return dict(case=case,parameter=parameter,secret=secret,next=advance,reset=reset_button,all=all_button,inspect=inspect_button,state=state)
```

Our running system stores highly confidential documents for twenty years. A well-funded adversary may steal storage, compromise a workload, misuse administrator access or retain old backups. The design must specify what happens after each event, not merely choose AES.

## A lifecycle is a set of permissions


```python
#@title Display teaching diagram
from IPython.display import display as _diagram_display, Image as _DiagramImage
import base64 as _diagram_base64
_diagram_display(_DiagramImage(data=_diagram_base64.b64decode('')))
```

Read the arrows as controlled transitions. A retired encryption key may remain necessary to read old data. A compromised signing key is not made trustworthy by renaming it retired. Define allowed operations, owners and evidence at each transition. Expiration, disabling use, revocation of trust and destruction are different actions.

| Decision | Record before deploying |
| --- | --- |
| Generation | Approved source of randomness or service, algorithm, purpose, identifier and owner |
| Distribution | Who receives raw material or permission to invoke operations, and through what authenticated path |
| Storage/use | Export policy, workload identity, tenant scope, operation limits and audit trail |
| Rotation | Trigger, new-write switch, old-read policy, migration progress and rollback constraints |
| Recovery | Which backups contain keys or wrapped keys, who can restore them, and tested recovery objectives |
| Destruction | All copies/dependencies, retention obligations, approval and evidence of removal |

Derivation can separate purposes, but compromise of a common parent can expose its descendants. Separate keys by purpose and appropriate tenant/data boundary; do not claim that many derived names provide independent custody.

## Envelope encryption

Encrypt each object with a DEK, then wrap that DEK using a KEK. Store the ciphertext and wrapped DEK with enough authenticated format/context information to recover them. A KMS may perform wrapping or unwrapping under access policy. Envelope encryption is a pattern, not proof that plaintext or DEKs never enter application memory.


```python
#@title Display teaching diagram
from IPython.display import display as _diagram_display, Image as _DiagramImage
import base64 as _diagram_base64
_diagram_display(_DiagramImage(data=_diagram_base64.b64decode('')))
```

We use AES Key Wrap from the library for a 32-byte DEK and AES-GCM for data. AES-KW supplies its own integrity mechanism and does not accept AAD in this API. We bind tenant/object context to the data with AEAD. This minimal envelope omits a production wire parser, authenticated routing metadata and remote KMS. See [NIST SP 800-38F](https://csrc.nist.gov/pubs/sp/800/38/f/final) and the [key-wrap API](https://cryptography.io/en/stable/hazmat/primitives/keywrap/).


```python
import secrets
from cryptography.hazmat.primitives.ciphers.aead import AESGCM
from cryptography.hazmat.primitives.keywrap import aes_key_wrap, aes_key_unwrap, InvalidUnwrap
from cryptography.exceptions import InvalidTag

def require_rejection(operation, exception):
    try:
        operation()
    except exception:
        return
    raise AssertionError('Expected rejection')

kek_old = secrets.token_bytes(32)
kek_new = secrets.token_bytes(32)
dek = secrets.token_bytes(32)
nonce = secrets.token_bytes(12)
context = b'archive:v1|tenant=acme|object=7'
plaintext = b'synthetic highly confidential archive document'
ciphertext = AESGCM(dek).encrypt(nonce, plaintext, context)
wrapped_old = aes_key_wrap(kek_old, dek)
recovered_dek = aes_key_unwrap(kek_old, wrapped_old)
assert AESGCM(recovered_dek).decrypt(nonce, ciphertext, context) == plaintext
require_rejection(lambda: aes_key_unwrap(kek_new, wrapped_old), InvalidUnwrap)
require_rejection(lambda: AESGCM(recovered_dek).decrypt(nonce, ciphertext, context + b'!'), InvalidTag)
print('PASS: envelope round trip; wrong KEK and changed context rejected')
```

The expected tenant/object context must come from the authorized request, not solely from attacker-controlled storage. In deployment, constrain key identifiers to an approved registry and bind meaningful metadata under the specified format. A copied record must not grant access just because its own AAD says another tenant.

## Rewrap is not re-encrypt


```python
wrapped_new = aes_key_wrap(kek_new, aes_key_unwrap(kek_old, wrapped_old))
assert aes_key_unwrap(kek_new, wrapped_new) == dek
assert AESGCM(aes_key_unwrap(kek_new, wrapped_new)).decrypt(nonce, ciphertext, context) == plaintext
# An attacker with the old KEK AND an old wrapped-DEK copy still recovers this DEK.
assert aes_key_unwrap(kek_old, wrapped_old) == dek
print('PASS: rewrap changes KEK protection, not the data ciphertext or historical copies')
```

Routine KEK rotation can rewrap DEKs without rewriting bulk data. If the DEK itself was exposed, generate a new DEK and re-encrypt the data with a new nonce. Even re-encryption cannot recall plaintext or ciphertext/key copies already held by the attacker.


```python
#@title Display teaching diagram
from IPython.display import display as _diagram_display, Image as _DiagramImage
import base64 as _diagram_base64
_diagram_display(_DiagramImage(data=_diagram_base64.b64decode('')))
```

Do not destroy the old KEK before all required retained data and backups are recoverable under the new design. Conversely, preserving every old key forever may preserve an adversary's recovery path. Recovery and destruction require a documented decision about retained information, not a universal fixed interval.

## Recovery and separation of duties

Test recovery from a clean environment using realistic key metadata and permissions, not only from a machine with cached credentials. Keep encrypted backups and their recovery authority separate enough to meet the threat model. Record recovery time and data-loss objectives, and verify actual restores against them.

Separate key administration, application use, security review and recovery approval where feasible. Two people sharing one unrestricted administrator account do not provide meaningful separation. Deleting a database row or a Python variable is not secure erasure; cryptographic erasure claims depend on eliminating all usable key copies and considering plaintext caches.
